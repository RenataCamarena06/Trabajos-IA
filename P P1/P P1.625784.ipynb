<h1 style="text-align: center;">Análisis de factores asociados al consumo excesivo de alcohol por entidad federativa en México</h1>

## Introducción
El consumo excesivo de alcohol es un problema relevante de salud pública en México. De acuerdo con la ENSANUT Continua 2023, el 41.9% de los adultos de 20 años o más reportó consumo excesivo de alcohol durante el último año, esta cifra representa alrededor de 36 millones de personas. Además, el 19.3% reportó consumo excesivo durante los 30 días antes de la encuesta. Sin embargo, también cabe destacar que estos números varían de distintas maneras en los 32 estados de la república mexicana.

Ante la magnitud de esta problemática, es importante analizar las características demográficas y de salud de la población que podrían estar asociadas con el consumo excesivo de alcohol. En este proyecto se estudiará su relación con la prevalencia de tabaquismo, la estructura de edad de la piblación, la prevalencia de sobrepeso u obesidad y la prevalencia de enfermedades en el país. 

El objetivo no es establecer que las variables previamente mencionadas sean causas directas del consumo excesivo de alcohol, sino determinar en qué medida su comportamiento permite expicar las diferencias observadas entre los estados de México. 

A partir de este problema, se plantea la siguiente pregunta de investigación: ¿En qué medida la prevalencia de tabaquismo, la estructura de edad de la población, la prevalencia de sobrepeso u obesidad y la prevalencia de enfermos permiten explicar las diferencias estatales en el consumo excesivo de alcohol en México?

La variable respuesta del estudio será la prevalencia de consumo excesivo de alcohol, mientras que la prevalencia de tabaquismo, la estructura de edad, la prevalencia de enfermedades y la prevalencia de sobrepeso u obesidad serán inicialmente consideradas como posibles variables explicativas. Asimismo, debido a que la variable de interés es cuantitativa y se busca analizar su relación con distintas características, se utilizarán modelos de regresión.

Este análisis permitirá evaluar la capacidad de estas variables para predecir las diferencias estatales en el consumo excesivo de alcohol y estudiarlas, usando inferencia estadística, para saber cuáles de ellas presentan una asociación significativa con la respuesta. Para lograr esto, se compararán modelos de regresión lineales y no lineales, tomando en consideración su desempeño predictivio y la interpretación de las relaciones encontradas.

## Descripción de los datos y fuentes de información
Para realizar este análisis se integró información proveniente de fuentes públicas y oficiales del gobierno de México. Las variables relacionadas con salud y hábitos de la población (tabaquismo, sobrepeso u obesidad y consumo excesivo de alcohol) fueron obtenidas principalmente de la Encuesta Nacional de Salud y Nutrición (ENSANUT), elaborada por el Instituto Nacional de Salud Pública. Esta encuesta tiene como objetivo actualizar información sobre las condiciones de salud y nutrición de la población de México y generar estimaciones para las 32 entidades federativas.

Para representar la estructura de edad de la población se utilizaron datos del Instituto Nacional de Estadpistica y Geografía (INEGI), esoecíficamente información de población por entidad federativa y grupos de edad. El INEGI tiene información por grupos quinquenales y entidad federativa, lo que permite calcular indicadores relacionados con la composición de edad de cada estado.

Adicionalmente, se consultó información del Anuario de Morbilidad de la Dirección General de Epidemiología de la Secretaría de Salud, donde se concentran registros de morbilidad por entidad federativa y permite incorporar información relacionada con la presencia de enfermedades en la población mexicana.

Después de integrar y procesar las diferentes fuentes de información, se construyó una base de datos en la que cada observaión representa una entidad federativa de México. Las variables utilizadas en el análisis son:

## Preparación de los datos
Durante la construcción del conjunto de datos se presentaron distintos problemas que requirieron un proceso de preparación antes del desarrollo de los modelos de regresión. Estos problemas estuvieron relacionados principalmente con el cálculo de prevalencias, la presencia de algunos valores atípicos y la posible colinealidad entre las variables explicativas.

### Cálculo e integración de las prevalencias
Uno de los principales retos fue que algunas de las fuentes consultadas no proporcionaban directamente las variables en forma de prevalencia. En ciertos casos se contaba solo con la cantidad de personas que presentaban una determinada característica.

Para poder comparar adecuadamente las entidades federativas, fue necesario complementar esta información con una base de datos de población del INEGI. A partir de ambas fuentes se calculó la prevalencia correspondiente para cada estado, relacionando el número de personas con la característica de interés respecto a la población correspondiente.

De manera general, las prevalencias fueron calculadas mediante:
$$
\text{Prevalencia} =
\frac{\text{Número de personas con la característica}}
{\text{Población correspondiente}}
$$

Este procedimiento permitió estandarizar la información entre entidades, ya que utilizar solamente el número de casos podría generar interpretaciones incorrectas debido a las diferencias en tamaño poblacional entre los estados.

### Revisión de valores atípicos
Posteriormente, se realizó una revisión de posibles valores atípicos en las variables utilizadas. Se identificaron algunas observaciones que presentaban valores más alejados del comportamiento general del conjunto de datos, esto se observa en la Tabla 4.

Sin embargo, al revisar estos casos individualmente se encontró que los valores se encontraban dentro de rangos completamente válidos y posibles para las variables analizadas. Por esta razón, se consideró que estas observaciones representan diferencias reales entre las entidades federativas y no errores en la recopilación de los datos.

Por lo tanto, se decidió conservar los valores atípicos dentro del conjunto de datos, para así evitar eliminar información que pudiera representar características reales y válidas de determinados estados. 

```python
resultados_outliers = []

for variable in variables_X:
    Q1 = train[variable].quantile(0.25)
    Q3 = train[variable].quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    outliers = train[(train[variable] < limite_inferior) | (train[variable] > limite_superior)]
    entidades = ", ".join(outliers["Entidad"].astype(str))
    resultados_outliers.append({"Variable": variable, "Límite inferior": limite_inferior,"Límite superior": limite_superior,"Cantidad de outliers": len(outliers),"Entidades identificadas": entidades if len(outliers) > 0 else "Ninguna"})

tabla_outliers = pd.DataFrame(resultados_outliers)
(tabla_outliers.style.hide(axis="index").format({"Límite inferior": "{:.3f}","Límite superior": "{:.3f}"}).set_properties(**{'text-align': 'center'}).set_table_styles([{'selector': 'th', 'props': [('text-align', 'center')]}]).set_caption("[TABLA 4]. Posibles valores atípicos en las variables predictoras del conjunto de entrenamiento mediante Tukey."))
```

### Revisión de colinealidad y separación de los datos
Antes de construir los modelos, el conjunto de datos se dividió en 70% para entrenamiento y 30% para prueba. Esta separación permite evaluar el desempeño del modelo con datos que no fueron utilizados durante su ajuste y ayuda a evitar fuga de datos.

Posteriormente, utilizando únicamente le conjunto de entrenamiento, se calculó la matriz de correlación para analizar la relación entre las variables y detectar posibles problemas de colinealidad. La colinealidad ocurre cuando dos o más variables explicativas tienen una relación elevada entre sí, lo que hace difícil dificultar la interpretación independiente de sus efectos en un modelo de regresión. No se encontraron correlaciones suficientemente altas entre las variables predictoras como para considerar que existiera un problema importante, por lo que se conservaron para continuar con la selección de características, esto se observa en la Gráfica 1.

```python
np.random.seed(34)

indices = np.random.permutation(len(data))
corte = int(len(data) * 0.70)

indices_train = indices[:corte]
indices_test = indices[corte:]

train = data.iloc[indices_train].copy()
test = data.iloc[indices_test].copy()

print("Entrenamiento:", train.shape)
print("Prueba:", test.shape)
```

## Selección de características
Con el objetivo de conservar únicamente las variables que aportan información relevante para explicar el consumo excesivo de alcohol, se aplicaron dos métodos de selección de características: selección hacia adelante y selección hacia atrás. El propósito de estos métodos es reducir el número de variables y facilitar la interpretación del modelo. Para realizar estos métodos de selección, se utilizó un nivel de significancia de 0.05. 

En la selección hacia adelante, la primera variable incorporada fue PrevalenciaFumadores, con un p-value de 0.0037. Al evaluar la incorporación de una segunda variable, ninguna presentó significancia estadística, por lo que el proceso se detuvo. Esto se observa en la Tabla 5.

Por otro lado, la selección hacia atrás comenzó considerando todas las variables y eliminó progresivamente aquellas con un p-value mayor a 0.05. Finalmente, este método también conservó únicamente PrevalenciaFumadores, con un p-value de 0.0037, esto es claramente representado en la Tabla 6.

En la Tabla 7, se observa que ambos métodos llegaron al mismo resultado, por lo que se seleccionó la prevalencia de fumadores como variable explicativa para la construcción de los modelos de regresión.

```python
def seleccion_adelante(data, variables_X, variable_Y, alpha=0.05):
    seleccionadas = []
    restantes = variables_X.copy()
    historial = []

    while len(restantes) > 0:
        resultados = []

        for variable in restantes:
            variables_prueba = seleccionadas + [variable]
            X = sm.add_constant(data[variables_prueba],has_constant="add")
            y = data[variable_Y]
            modelo = sm.OLS(y, X).fit()
            resultados.append({"Variable": variable,"p-value": modelo.pvalues[variable]})

        tabla_temporal = pd.DataFrame(resultados)
        mejor = tabla_temporal.loc[tabla_temporal["p-value"].idxmin()]
        mejor_variable = mejor["Variable"]
        mejor_p = mejor["p-value"]
        historial.append({"Variable evaluada": mejor_variable,"p-value": mejor_p})

        if mejor_p < alpha:
            seleccionadas.append(mejor_variable)
            restantes.remove(mejor_variable)
        else:
            break

    return seleccionadas, pd.DataFrame(historial)


variables_forward, historial_forward = seleccion_adelante(train,variables_X,variable_Y)

(historial_forward.style.hide(axis="index").format({"p-value": "{:.4f}"}).set_properties(**{'text-align': 'center'}).set_table_styles([{'selector': 'th', 'props': [('text-align', 'center')]}]).set_caption("[TABLA 5]. Selección de características hacia adelante."))
```

```python
def seleccion_atras(data, variables_X, variable_Y, alpha=0.05):
    seleccionadas = variables_X.copy()
    historial = []

    while len(seleccionadas) > 0:
        X = sm.add_constant(data[seleccionadas],has_constant="add")
        y = data[variable_Y]
        modelo = sm.OLS(y, X).fit()
        p_values = modelo.pvalues.drop("const")
        peor_variable = p_values.idxmax()
        peor_p = p_values.max()
        historial.append({"Variable evaluada": peor_variable,"p-value": peor_p})

        if peor_p > alpha:
            seleccionadas.remove(peor_variable)
        else:
            break

    return seleccionadas, pd.DataFrame(historial)


variables_backward, historial_backward = seleccion_atras(train,variables_X,variable_Y)
(historial_backward.style.hide(axis="index").format({"p-value": "{:.4f}"}).set_properties(**{'text-align': 'center'}).set_table_styles([{'selector': 'th', 'props': [('text-align', 'center')]}]).set_caption("[TABLA 6]. Selección de características hacia atrás."))
```

```python
comparacion_seleccion = pd.DataFrame({"Método": ["Selección hacia adelante","Selección hacia atrás"],"Variables seleccionadas": [", ".join(variables_forward) if variables_forward else "Ninguna",", ".join(variables_backward) if variables_backward else "Ninguna"],"Cantidad de variables": [len(variables_forward),len(variables_backward)]})
(comparacion_seleccion.style.hide(axis="index").set_properties(**{'text-align': 'center'}).set_table_styles([{'selector': 'th', 'props': [('text-align', 'center')]}]).set_caption("[TABLA 7]. Comparación de los métodos de selección paso a paso."))
```

## Construcción del modelo lineal
A partir de las variables seleccionadas, se construyó un modelo de regresión lineal simple utilizando únicamente el conjunto de entrenamiento. El objetivo de este modelo es analizar la relación entre la variable seleccionada y la prevalencia de consumo excesivo de alcohol. Esto también tiene el propósito de establecer un modelo base para posteriormente compararlo con una alternativa no lineal.

En la Tabla 8, se pueden observar los coeficientes estimados de las variables, los p-values y otros estadísticos, cómo el error estándar y los intervalos de confianza. Los coeficientes estimados permiten conocer la magnitud de la relación de cada variable con la respuesta, mientras que los p-values permiten evaluar si dicha asociación es estadísticamente significativa.

El coeficiente obtenido para la prevalencia de fumadores fue de 0.7329, lo que indica una relación positiva con la prevalencia de consumo excesivo de alcohol. Debido a que ambas variables están expresadas como proporciones, un aumento de 1 punto porcentual en la prevalencia de fumadores se asocia con un aumento de aproximadamente 0.73 puntos porcentuales en la prevalencia de consumo excesivo de alcohol.

```python
X_train_modelo = sm.add_constant(train[variables_backward],has_constant="add")
X_test_modelo = sm.add_constant(test[variables_backward],has_constant="add")
y_train = train[variable_Y]
y_test = test[variable_Y]
modelo = sm.OLS(y_train,X_train_modelo).fit()

tabla_coeficientes = pd.DataFrame({"Variable": modelo.params.index,"Coeficiente": modelo.params.values,"Error estándar": modelo.bse.values,"Estadístico t": modelo.tvalues.values,"p-value": modelo.pvalues.values,"IC 95% inferior": modelo.conf_int()[0].values,"IC 95% superior": modelo.conf_int()[1].values})
(tabla_coeficientes.style.hide(axis="index").format({"Coeficiente": "{:.4f}","Error estándar": "{:.4f}","Estadístico t": "{:.4f}","p-value": "{:.4f}","IC 95% inferior": "{:.4f}","IC 95% superior": "{:.4f}"}).set_properties(**{"text-align": "center"}).set_table_styles([{"selector": "th","props": [("text-align", "center")]}]).set_caption("[TABLA 8]. Coeficientes y significancia estadística del modelo de regresión lineal simple con los datos de entrenamiento."))
```

### Evaluación del modelo lineal
De acuerdo con la Tabla 9, el modelo obtuvo un R² de 0.3498, lo que indica que la prevalencia de fumadores permite explicar aproximadamente el 35 % de la variabilidad observada en la prevalencia de consumo excesivo de alcohol dentro de los datos de entrenamiento. El R² ajustado fue de 0.3172, un valor cercano al R² original.

El modelo también presentó un estadístico F de 10.7578 con un p-value de 0.0037. Debido a que este valor es menor a 0.05, existe evidencia estadística de que el modelo de regresión es significativo, por lo que la relación encontrada entre la prevalencia de fumadores y el consumo excesivo de alcohol es relevante.

En cuanto al error de predicción, el modelo obtuvo un RMSE de entrenamiento de 0.0343, equivalente a aproximadamente 3.43 puntos porcentuales. Esto significa que las predicciones realizadas sobre los datos utilizados para construir el modelo presentan errores relativamente pequeños respecto a los valores reales.

```python
pred_train = modelo.predict(X_train_modelo)
rmse_train = np.sqrt(np.mean((y_train - pred_train)**2))
rss = np.sum((y_train - pred_train)**2)

tabla_modelo = pd.DataFrame({"R²": [modelo.rsquared],"R² ajustada": [modelo.rsquared_adj],"Estadístico F": [modelo.fvalue],"p-value F": [modelo.f_pvalue],"RSS": [rss],"RMSE Train": [rmse_train]})
(tabla_modelo.style.hide(axis="index").format({"R²": "{:.4f}","R² ajustada": "{:.4f}","Estadístico F": "{:.4f}","p-value F": "{:.4f}","RSS": "{:.4f}","RMSE Train": "{:.4f}"}).set_properties(**{"text-align": "center"}).set_table_styles([{"selector": "th","props": [("text-align", "center")]}]).set_caption("[TABLA 9]. Estadísticos generales del modelo de regresión lineal simple con datos de entrenamiento."))
```

La Tabla 10 muestra que, al evaluar el modelo con datos de prueba, se obtuvo un R² de 0.4307, por lo que el modelo explicó aproximadamente el 43.1 % de la variabilidad de estas observaciones, mientras que el RMSE aumentó ligeramente de 3.43 a 4.12 puntos porcentuales.

En general, la diferencia entre los errores de entrenamiento y prueba es relativamente pequeña, por lo que no hay alguna indicación de un sobreajuste considerable. El modelo mantiene un comportamiento similar frente a observaciones que no fueron utilizadas durante su entrenamiento. Aunque todavía existe una proporción considerable de la variabilidad del consumo excesivo de alcohol que no es explicada únicamente por la prevalencia de tabaquismo.

```python
pred_test = modelo.predict(X_test_modelo)

r2_train = modelo.rsquared
r2_ajustado_train = modelo.rsquared_adj

ss_res_test = np.sum((y_test - pred_test)**2)
ss_tot_test = np.sum((y_test - y_test.mean())**2)
r2_test = 1 - (ss_res_test / ss_tot_test)

rmse_test = np.sqrt(np.mean((y_test - pred_test)**2))

mae_train = np.mean(np.abs(y_train - pred_train))

mae_test = np.mean(np.abs(y_test - pred_test))

tabla_desempeno = pd.DataFrame({"Modelo": ["Selección hacia atrás"],"R² Train": [r2_train],"R² ajustada Train": [r2_ajustado_train],"R² Test": [r2_test],"RMSE Train": [rmse_train],"RMSE Test": [rmse_test],"MAE Train": [mae_train],"MAE Test": [mae_test]})
(tabla_desempeno.style.hide(axis="index").format({"R² Train": "{:.4f}","R² ajustada Train": "{:.4f}","R² Test": "{:.4f}","RMSE Train": "{:.4f}","RMSE Test": "{:.4f}","MAE Train": "{:.4f}","MAE Test": "{:.4f}"}).set_properties(**{"text-align": "center"}).set_table_styles([{"selector": "th","props": [("text-align", "center")]}]).set_caption("[TABLA 10]. Comparación del desempeño del modelo de regresión lineal simple en datos de entrenamiento y prueba."))
```

También se observa que algunos estados se alejan más de la línea estimada, lo que indica que el modelo no logra explicar por completo todas las diferencias estatales. Esto sugiere que, además del tabaquismo, existen otros factores que podrían estar relacionados con el consumo excesivo de alcohol.

## Construcción del modelo no lineal
Para evaluar si la relación entre la prevalencia de fumadores y el consumo excesivo de alcohol podía representarse mejor mediante una relación no lineal, se construyó un modelo de regresión polinomial de grado 2.

Este tipo de modelo agrega un término cuadrático de la variable explicativa, permitiendo que la relación pueda presentar cierta curvatura en lugar de asumir que el cambio es completamente constante en todo el rango de los datos.

El modelo utilizado puede expresarse como:
$$
\hat{Y} = \hat{\beta}_0 + \hat{\beta}_1 X + \hat{\beta}_2 X^2
$$

donde X corresponde a la prevalencia de fumadores y Y a la prevalencia estimada de consumo excesivo de alcohol.

El modelo fue ajustado utilizando el mismo conjunto de entrenamiento empleado en la regresión lineal y posteriormente evaluado con el mismo conjunto de prueba. De esta manera, ambos modelos pueden compararse bajo las mismas condiciones y determinar si la incorporación del término cuadrático representa una mejora real en la capacidad de explicar y predecir el consumo excesivo de alcohol.

La Tabla 11 muestra los coeficientes estimados para el modelo polinomial de grado 2. El término lineal de PrevalenciaFumadores obtuvo un coeficiente de 2.5950, mientras que el término cuadrático presentó un coeficiente de −5.2577. En conjunto, estos valores describen que el consumo excesivo de alcohol tiende inicialmente a aumentar conforme aumenta la prevalencia de fumadores, pero este crecimiento se va reduciendo en los valores más altos.

Sin embargo, los p-values del término lineal (0.2318) y del término cuadrático (0.3839) son mayores a 0.05, por lo que ninguno presenta significancia estadística de manera individual.

Por lo tanto, aunque el modelo de grado 2 permite representar una relación más flexible, no existe evidencia estadística suficiente para afirmar que el término cuadrático aporta una mejora significativa en la relación entre tabaquismo y consumo excesivo de alcohol.

```python
variable_simple = variables_backward[0]
X_train_modelo = pd.DataFrame({variable_simple: train[variable_simple],f"{variable_simple}^2": train[variable_simple]**2})
X_test_modelo = pd.DataFrame({variable_simple: test[variable_simple],f"{variable_simple}^2": test[variable_simple]**2})
X_train_modelo = sm.add_constant(X_train_modelo, has_constant="add")
X_test_modelo = sm.add_constant(X_test_modelo, has_constant="add")
y_train = train[variable_Y]
y_test = test[variable_Y]
modelo = sm.OLS(y_train, X_train_modelo).fit()

tabla_coeficientes = pd.DataFrame({"Variable": modelo.params.index,"Coeficiente": modelo.params.values,"Error estándar": modelo.bse.values,"Estadístico t": modelo.tvalues.values,"p-value": modelo.pvalues.values,"IC 95% inferior": modelo.conf_int()[0].values,"IC 95% superior": modelo.conf_int()[1].values})
(tabla_coeficientes.style.hide(axis="index").format({"Coeficiente": "{:.4f}","Error estándar": "{:.4f}","Estadístico t": "{:.4f}","p-value": "{:.4f}","IC 95% inferior": "{:.4f}","IC 95% superior": "{:.4f}"}).set_properties(**{"text-align": "center"}).set_table_styles([{"selector": "th", "props": [("text-align", "center")]}]).set_caption("[TABLA 11]. Coeficientes y significancia estadística del modelo de regresión polinomial de grado 2."))
```

Al considerar la complejidad adicional del modelo, el R² ajustado disminuye a 0.3102, lo que demuestra que incluir un polinomio de grado 2 no es la mejor alternativa.

El modelo en conjunto presenta un p-value F de 0.0114, menor a 0.05, por lo que puede considerarse estadísticamente significativo. Esto significa que, en conjunto, los términos incluidos aportan información para explicar la variable respuesta, aunque anteriormente se observó que los coeficientes individuales del término lineal y cuadrático no resultaron significativos. En cuanto al error, el RMSE de entrenamiento fue de 0.0336, equivalente aproximadamente a 3.36 puntos porcentuales.

## Conclusiones
El objetivo de este proyecto fue analizar en qué medida la prevalencia de tabaquismo, la estructura de edad, la prevalencia de sobrepeso u obesidad y la prevalencia de enfermos permiten explicar las diferencias estatales en el consumo excesivo de alcohol en México.

Los métodos de selección de características hacia adelante y hacia atrás coincidieron en conservar únicamente la prevalencia de fumadores. Las demás variables no mostraron evidencia estadística suficiente para mantenerse dentro del modelo final. Para el tabaquismo se obtuvo un coeficiente de 0.7329 y un p-value de 0.0037, lo que indica que un aumento de un punto porcentual en la prevalencia de fumadores se asocia, en promedio, con un incremento aproximado de 0.73 puntos porcentuales en el consumo excesivo de alcohol.

Además, la incorporación del término cuadrático no produjo una mejora estadísticamente significativa (p=0.3839), por lo que se seleccionó el modelo lineal por su menor complejidad y mejor interpretación.

Por lo tanto, la pregunta de investigación se responde de manera parcial. Los resultados no respaldan que las cuatro variables expliquen conjuntamente las diferencias estatales en el consumo excesivo de alcohol.

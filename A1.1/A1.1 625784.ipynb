```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("A1.1 Obesidad.csv")
print(data.head())
```

### Propuesta de variable adicional
Además de las variables ya incluidas en la base de datos, sería útil incorporar una variable que permita describir con mayor detalle la composición corporal de los participantes, por ejemplo, la medida de la circunferencia abdominal. Esto ayudaría a complementar la información relacionada con el peso y los niveles de obesidad, ya que aportaría una perspectiva adicional sobre la distribución de grasa corporal.

Esta variable es de naturaleza cuantitativa, debido a que se expresa mediante una medida numérica (centímetros). Su inclusión sería relevante porque existen valores de referencia distintos para hombres y mujeres que permiten identificar cuándo una circunferencia abdominal puede asociarse con una mayor acumulación de grasa y, por lo tanto, con un mayor riesgo relacionado con la obesidad. De esta manera, esta variable aportaría información adicional a medidas como el peso o el índice de masa corporal, ya que permitiría analizar no solo cuánto pesa una persona, sino también cómo se distribuye parte de esa grasa corporal.

### Variable de inferencia
Para el estudio de inferencia se seleccionó la variable ComeMuchasCalorias como variable de salida. Esta elección se realizó debido a que el consumo frecuente de alimentos con alto contenido calórico representa un hábito alimenticio relevante dentro del estudio de la obesidad y permite analizar su posible asociación con otras características de los individuos.

La variable es de naturaleza cualitativa, ya que clasifica a los participantes en dos categorías según consuman o no un alto contenido calórico. Por esta razón, su análisis descriptivo se puede realizar mediante la frecuencia y el porcentaje de individuos pertenecientes a cada categoría, lo que permite conocer cómo se distribuye este hábito dentro de la población estudiada.

Para realizar un análisis descriptivo básico de esta variable se utilizó la función value_counts() en Python, con el objetivo de conocer cuántas personas pertenecían a cada una de las categorías. Esto permitió observar la distribución de los participantes entre quienes reportaron consumir muchas calorías y quienes no, facilitando una primera interpretación de la frecuencia de este hábito dentro de la población estudiada.

```python
data["ComeMuchasCalorias"].value_counts()
```

### Resultados
Una vez realizada la exploración de los datos y definidas las variables de interés, se analizó la relación entre el consumo de muchas calorías y el consumo de vegetales dentro de la población masculina. Para ello, se compararon los valores de ComeVegetales entre los hombres que reportaron consumir muchas calorías y aquellos que no.

En la gráfica se observa que los hombres que no consumen muchas calorías presentan, en general, un mayor consumo de vegetales. En este grupo, una buena parte de los datos se encuentra entre valores de 2 y 3, mientras que en los hombres que sí consumen muchas calorías los datos se concentran principalmente alrededor de valores cercanos a 2, aunque también existen personas con un consumo alto de vegetales.

También se puede observar que ambos grupos presentan valores desde aproximadamente 1 hasta 3, por lo que existe cierta similitud entre ellos. Sin embargo, la distribución muestra que el grupo que no consume muchas calorías tiende a tener valores más altos de consumo de vegetales. Esto permite observar una posible relación entre ambas variables, donde un menor consumo de muchas calorías parece estar acompañado por un mayor consumo de vegetales dentro de la población masculina estudiada.

```python
import matplotlib.pyplot as plt

# Seleccionar únicamente hombres
hombres = data[data["Sexo"] == "Male"]

# Crear gráfica
hombres.boxplot(
    column="ComeVegetales",
    by="ComeMuchasCalorias",
    figsize=(8, 5)
)

plt.title("Consumo de vegetales según el consumo de muchas calorías en hombres")
plt.suptitle("")
plt.xlabel("Consume muchas calorías")
plt.ylabel("Consumo de vegetales")
plt.xticks([1, 2], ["No", "Sí"])

plt.show()
```

### Conclusión
El análisis realizado permitió identificar una posible relación entre distintos hábitos alimenticios dentro de la población estudiada. En particular, los resultados sugieren que un mayor consumo de vegetales podría estar asociado con un menor consumo de alimentos con muchas calorías en la población masculina. Sin embargo, las diferencias observadas no son suficientes para afirmar que esta relación se presente de la misma manera en todos los individuos.

En conjunto, este ejercicio permitió mostrar la utilidad del análisis estadístico para encontrar patrones y relaciones entre variables relacionadas con la obesidad. También deja claro que este tipo de resultados deben interpretarse con cuidado, ya que una asociación entre variables no implica necesariamente una relación de causa y efecto.
